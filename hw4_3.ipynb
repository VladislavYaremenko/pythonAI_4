```python
import matplotlib.pyplot as plt

group1 = [60, 65, 70, 72, 75, 80, 85]
group2 = [55, 60, 68, 75, 82, 90, 95]
group3 = [70, 72, 74, 76, 78, 80, 82]

plt.boxplot(
    [group1, group2, group3],
    tick_labels=["Група 1", "Група 2", "Група 3"]
)

plt.xlabel("Групи студентів")
plt.ylabel("Оцінки")
plt.title("Порівняння результатів іспитів")
plt.show()
```
