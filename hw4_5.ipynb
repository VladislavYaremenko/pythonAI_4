```python
import matplotlib.pyplot as plt

hours = list(range(24))
load = [20, 15, 12, 10, 10, 15, 25, 40,
        55, 65, 70, 75, 80, 85, 80, 75,
        70, 85, 95, 90, 75, 60, 45, 30]

plt.plot(hours, load, marker="o")
plt.fill_between(hours, load, alpha=0.3)

plt.xlabel("Година")
plt.ylabel("Навантаження сервера (%)")
plt.title("Навантаження сервера протягом доби")
plt.grid(True)
plt.show()
```
